# Euler's Formula — Power Series Construction

The Taylor series of $e^{i\varphi}$ is:

$$e^{i\varphi} = \sum_{k=0}^{\infty} \frac{(i\varphi)^k}{k!} = 1 + i\varphi + \frac{(i\varphi)^2}{2!} + \cdots$$

Each term is a complex number.  Laid **head-to-tail**, they form a vector chain
that spirals toward the exact point $e^{i\varphi}$ on the unit circle.

Use the slider to change $\varphi$.

```python
import numpy as np
import ipywidgets as widgets
import matplotlib.pyplot as plt
from cnlecture.visualizations.euler import euler_series_terms, euler_series_points
```

```python
%matplotlib widget

N_TERMS = 30

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect('equal')
ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(1, 0, '>k', transform=ax.get_yaxis_transform(), clip_on=False)
ax.plot(0, 1, '^k', transform=ax.get_xaxis_transform(), clip_on=False)
ax.grid(zorder=0)

t_circ = np.linspace(0, 2 * np.pi, 200)


def draw(phi: float) -> None:
    """Redraw the spiral for the given phi."""
    for artist in ax.lines + ax.collections:
        artist.remove()

    # Unit circle
    ax.plot(np.cos(t_circ), np.sin(t_circ), 'C2-', lw=1, label='Unit circle')

    # Exact target point
    exact = np.exp(1j * phi)
    ax.plot(
        exact.real, exact.imag,
        'x', ms=14, mew=3, c='C0',
        label=rf'$e^{{i\cdot{phi/np.pi:.2f}\pi}}$',
    )

    # Vector chain from package helpers
    terms = euler_series_terms(phi, N_TERMS)
    points = euler_series_points(phi, N_TERMS)

    for k, term in enumerate(terms):
        tail = points[k - 1] if k > 0 else 0 + 0j
        ax.quiver(
            tail.real, tail.imag,
            term.real, term.imag,
            scale_units='xy', scale=1, width=0.01,
            color='C3', zorder=5,
        )

    # Axes limits
    lim = max(1.2, float(np.abs(terms).max()) + 0.2)
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
    fig.canvas.draw_idle()


draw(1.0)


@widgets.interact(phi=widgets.FloatSlider(
    value=1.0, min=0.0, max=2 * np.pi, step=0.1,
    description='φ (rad)', continuous_update=True,
))
def update(phi=1.0):
    draw(phi)
```
